```python
plt.figure()
sns.scatterplot(x=df['species'],y=df['sepal_length'])
plt.title('Scatter Plot')
plt.xlabel('species')
plt.ylabel('sepal_length')
plt.show()
```

```python
plt.figure()
sns.histplot(df['sepal_length'], bins=20, kde=True)
plt.title('Histogram')
plt.xlabel('sepal_length')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure()
sns.boxplot(x=df['sepal_length'])
plt.title('Box Plot')
plt.show()
```
